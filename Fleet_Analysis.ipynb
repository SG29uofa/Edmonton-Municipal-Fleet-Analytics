```python
fleet_summary = df.pivot_table(
    index=["department", "asset class"],
    columns=df["period"].dt.year,
    values="kilometers driven",
    aggfunc="sum"
)

fleet_summary.head(10)
```

```python
fleet_summary["km_change"] = (
    fleet_summary[2014] - fleet_summary[2012]
)
```

```python
fleet_summary.sort_values(
    "km_change", ascending=False
).head(10)
```

```python
fleet_summary["pct_change"] = (
    fleet_summary["km_change"] / fleet_summary[2012]
) * 100
```

```python
fleet_summary.sort_values(
    "km_change", ascending=False
).head(10)
```

```python
fleet_coverage = df.pivot_table(
    index=["department", "asset class"],
    columns=df["period"].dt.year,
    values="period",
    aggfunc="nunique"
)
```

```python
fleet_coverage.head(10)
```

```python
fleet_summary["months_2012"] = fleet_coverage[2012]
fleet_summary["months_2013"] = fleet_coverage[2013]
fleet_summary["months_2014"] = fleet_coverage[2014]
```

```python
fleet_summary.head(10)
```

```python
fleet_summary["coverage_status"] = (
    (fleet_summary["months_2012"] == 12) &
    (fleet_summary["months_2013"] == 12) &
    (fleet_summary["months_2014"] == 12)
)
```

```python
fleet_summary["coverage_status"].value_counts()
```

```python
fleet_complete = fleet_summary[
    fleet_summary["coverage_status"] == True
].copy()
```

```python
fleet_complete.shape
```

```python
fleet_complete["abs_km_change"] = (
    fleet_complete["km_change"].abs()
)
```

```python
fleet_complete["abs_pct_change"] = (
    fleet_complete["pct_change"].abs()
)
```

```python
fleet_complete[
    ["km_change", "pct_change", "abs_km_change", "abs_pct_change"]
].head(10)
```

```python
fleet_complete["priority_score"] = (
    fleet_complete["abs_km_change"].rank(pct=True) * 50
    +
    fleet_complete["abs_pct_change"].rank(pct=True) * 50
)
```

```python
fleet_complete.sort_values(
    "priority_score", ascending=False
).head(10)
```
